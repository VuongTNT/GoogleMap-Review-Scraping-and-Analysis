```python
if not os.path.exists(locs_directory):
    os.makedirs(locs_directory)
    print(f"📁 Created directory: {locs_directory}")
else:
    print(f"✅ Directory already exists: {locs_directory}")
```

```python
input_file = f"{locs_directory}/input.txt"

#Choose keywords for searching, mode="w" for overwriting, "a" for append new data
with open(input_file, "w", encoding="utf-8") as f:
    f.write("restaurant\n")
    f.write("bar\n")
    f.write("cafe\n")
    f.write("fastfood\n")
    f.write("food\n")
    f.write("beverage\n")
    f.write("noodles\n")

print(f"✅ Done writing to '{input_file}' ")
```

```python
def get_geo():

    GEO_BOUNDS = {
        "USA": {
            "latitude": (32, 49),
            "longitude": (-124, -66)   
        },
        "UK": {
            "latitude": (50, 58),     
            "longitude": (-5, 1)       
        },
        "VN_HN": {
            "latitude": (20, 22),     
            "longitude": (105, 107)      
        },
        "VN_HCM": {
            "latitude": (10, 11),     
            "longitude": (106, 108)      
        },
    }

    country_name = random.choice(list(GEO_BOUNDS.keys()))
    
    lat_min, lat_max = GEO_BOUNDS[country_name]["latitude"]
    lon_min, lon_max = GEO_BOUNDS[country_name]["longitude"]

    random_lat = random.uniform(lat_min, lat_max)
    random_lon = random.uniform(lon_min, lon_max)

    print(f"Country Chosen: {country_name}")
    print(f"Latitude Range: [{lat_min}° to {lat_max}°]")
    print(f"Longitude Range: [{lon_min}° to {lon_max}°]")
    
    return (random_lat, random_lon, country_name)
```

```python
#Run till you find it enough, then stop manually, remember to choose "Cancel" if you are using VSCode
count = 0 

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")    #Timestamp of when loop started
while count < 5: #Set number of loops
    print("Count: ", count)
    long, lat, country_name = get_geo()
    data_ids_output_file_name = f"{locs_directory}/{locs_output_file_name}-{timestamp}-{str(count)}-{country_name}.csv"
    geo = f"{str(long)},{str(lat)}"

    cmd = [
        SCRAPER_PATH,
        "-input", input_file,
        "-results", data_ids_output_file_name,
        "-exit-on-inactivity", "1m",
        "-fast-mode",
        "-geo", geo,
        "-radius", "500000", #Increase radius for faster search
        "-depth", "20"
    ]

    print(f"🚀 Running scraper for coordinates: {geo}")
    print(f"📄 Output: {data_ids_output_file_name}")

    result = subprocess.run(cmd, capture_output=True, text=True)
    print(f"✅ Finished for count {count} and coordinates {geo}\n\n\n")
    count += 1
```

## Extract data IDs

```python
data_ids_output_file_name = f"{locs_directory}/data_ids.csv"  

id_col = "data_id"          
review_col = "review_count"

locs_files = [f for f in os.listdir(locs_directory) if f.endswith(".csv") and f.startswith("locs-")]

if not locs_files:
    print(f"⚠️ No location file found in {locs_directory}")
else:
    print(f"🔍 Found {len(locs_files)} CSV files: {locs_files}")

unique_rows = dict()  # key = data_id, value = review_count

for csv_file in locs_files:
    file_path = os.path.join(locs_directory, csv_file)
    try:
        df = pd.read_csv(file_path)
        if id_col in df.columns and review_col in df.columns:
            for _, row in df.iterrows():
                data_id = str(row[id_col])
                review_count = row[review_col]
                if data_id not in unique_rows:
                    unique_rows[data_id] = review_count
            print(f"✅ Processed {csv_file}, total unique IDs so far: {len(unique_rows)}")
        else:
            print(f"⚠️ Skipped {csv_file}: missing '{id_col}' or '{review_col}' column")
    except Exception as e:
        print(f"❌ Error reading {csv_file}: {e}")

if unique_rows:
    output_df = pd.DataFrame(list(unique_rows.items()), columns=[id_col, review_col])
    output_df.to_csv(data_ids_output_file_name, index=False)
    print(f"🎉 Saved {len(unique_rows)} unique rows to {data_ids_output_file_name}")
else:
    print("⚠️ No rows found to save.")

```

### Find Duplication and Remove Duplicated Files

```python
from collections import defaultdict


review_map = defaultdict(list)

for filename in os.listdir(responses_directory):
        if not filename.endswith(".json"):
            continue

        file_path = os.path.join(responses_directory, filename)
        try:
            with open(file_path, "r", encoding="utf-8") as f:
                data = json.load(f)
        except Exception as e:
            print(f"⚠️ Could not read {file_path}: {e}")
            continue

        reviews = data.get("data", {}).get("reviews", [])
        if not reviews:
            continue

        for review in reviews:
            rid = review.get("review_id")
            if not rid:
                continue
            review_map[rid].append(file_path)

duplicates = {rid: files for rid, files in review_map.items() if len(files) > 1}

if duplicates:
    print("\n🚨 Duplicated review_ids found!")
    for rid, files in duplicates.items():
        print(f"\n🔁 Review ID: {rid}")
        for f in files:
            print(f"   → {f}")
    print(f"\n❌ Total duplicated review_ids: {len(duplicates)}")
else:
    print("✅ No duplicated review_ids found across all folders.")

print(f"📦 Total unique review_ids scanned: {len(review_map)}")
```

```python
#Remove duplication from group with more than 5 duplicated files 
dup_set = {rid: tuple(files) for rid, files in duplicates.items()}
dup_file = {dup_set[rid]: 0 for rid in dup_set}

for rid in dup_set:
    dup_file[dup_set[rid]] += 1

for files_tuple, count in dup_file.items():
    if count > 5:
        print(f"🗑 Removing files for group (count={count}):")
        for file_path in files_tuple[:-1]:  # keep the last file only
            try:
                os.remove(file_path)
                print(f"   → Deleted: {file_path}")
            except Exception as e:
                print(f"   ⚠️ Could not delete {file_path}: {e}")
        print(f"✅ Kept last file: {files_tuple[-1]}")

print("✨ Cleanup completed!")

```

```python
folder = "raw_data_from_api"  # 🔁 change to your folder path
error_msg = "You have exceeded the MONTHLY quota for Requests on your current plan"

found_files = []

for file in os.listdir(folder):
    if file.endswith(".json"):
        file_path = os.path.join(folder, file)
        try:
            with open(file_path, "r", encoding="utf-8") as f:
                data = json.load(f)
            
            # check if it has the message key 
            if isinstance(data, dict) and "message" in data:
                found_files.append(file)
        except Exception as e:
            print(f"⚠️ Could not read {file}: {e}")

if found_files:
    print(f"🚫 Found {len(found_files)} file(s) with quota error:")
    for f in found_files:
        print("  •", f)
else:
    print("✅ No files contain the quota error message.")
```

## Combine reviews and rating from json into a dataframe and save to csv

```python
#folder = "raw_data_from_api\\raw_data_from_api"
folder = responses_directory

extracted_data = {'review_id': [], 'text': [], 'rating': []}

filedirs = os.listdir(folder)
total_files = len(filedirs)
processed_files = 0
num_of_reviews = 0
for filename in filedirs:
    processed_files += 1
    print(f"Read {processed_files} of {total_files} files, added {num_of_reviews} reviews", end='\r')
    
    if not filename.endswith(".json"):
        continue

    file_path = os.path.join(folder, filename)

    try:
        with open(file_path, "r", encoding="utf-8") as f:
            data = json.load(f)
    except Exception as e:
        print(f"⚠️ Could not read {file_path}: {e}")
        continue

    reviews = data.get("data", {}).get("reviews", [])
    if not reviews:
        continue

    for review in reviews:
        review_id = review['review_id']
        rating = review['review_rate']
        
        translation = review.get("translation", {})
        text = translation.get('en') or review.get("review_text", "")
        if text is None: continue
        text = text.replace("\n", "") or ""      #remove line break so it can be save into a csv"
        if len(text.strip()) < 10:
            continue                            #reject too short reviews
        
        
        extracted_data['review_id'].append(review_id)
        extracted_data['text'].append(text)
        extracted_data['rating'].append(rating)
        num_of_reviews += 1

print()
df = pd.DataFrame(extracted_data)
print("Number of reviews: " + str(df.shape[0]))
df = df.drop_duplicates()
print("After dropping duplicates: " + str(df.shape[0]))
df.to_csv('reviews.csv', index=False)
```
