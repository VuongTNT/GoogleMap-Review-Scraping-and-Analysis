```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from joblib import load
from tabulate import tabulate
```

```python
from abc import abstractmethod

class Model:
    @abstractmethod    
    def predict(self, texts: list) -> list:
        pass
    
    @abstractmethod
    def grid_search(self, train_df):
        pass
```

```python
class SVMModel(Model):
    def __init__(self, svm: LinearSVC, vectorizer: TfidfVectorizer):
        self.svm = svm
        self.vec = vectorizer

    def predict(self, texts: list):
        vec = self.vec.transform(texts)
        pred_labels = self.svm.predict(vec)
        return pred_labels

    def __str__(self):
        res = f"SVM-Model[\n   {str(self.svm)},\n   {str(self.vec)}\n]"
        return res
```

```python
with open("../../../models/svm-tfidf-10k.pkl", "rb") as f:
    svm: LinearSVC = load(f)
with open("../../../models/tfidf-vectorize-10k.pkl", "rb") as f:
    vectorizer: TfidfVectorizer = load(f)
model = SVMModel(svm, vectorizer)
print(model)
```

```python
reviews = [
    "The food is great",
    "The food is ok",
    "The food is terrible",
]
p_ratings = model.predict(reviews).tolist()
print(tabulate(
    zip(reviews, p_ratings),
    headers=["Reviewss", "Predict Ratings"]))
```
